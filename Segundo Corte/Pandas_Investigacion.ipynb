```python
# Importar la librería Pandas
import pandas as pd

print("Pandas importado correctamente")
```

```python
# Ejercicio 1 - Crear un DataFrame
```

```python
# Crear datos
datos = {
    "Nombre": ["Ana", "Luis", "Carlos", "Marta"],
    "Edad": [20, 22, 21, 23],
    "Nota": [4.5, 3.8, 4.2, 4.9]
}

# Crear DataFrame
df = pd.DataFrame(datos)

# Mostrar datos
print(df)
```

```python
# Ejercicio 2 - Operaciones entre columnas
```

```python
# Crear nueva columna con aumento del 10%
df["Nota_Final"] = df["Nota"] * 1.10

print(df)
```

```python
# Ejercicio 3 - Calcular promedio
```

```python
promedio = df["Nota"].mean()

print("El promedio es:", promedio)
```

```python
# Ejercicio 4 - Datos sintéticos
```

```python
# Importar NumPy
import numpy as np

# Crear datos aleatorios
datos_random = pd.DataFrame({
    "Temperatura": np.random.randint(20, 35, 10),
    "Humedad": np.random.randint(50, 90, 10)
})

print(datos_random)
```

```python
# Ejercicio 5 - Guardar archivo CSV
```

```python
# Guardar CSV
df.to_csv("estudiantes.csv", index=False)

print("Archivo CSV guardado correctamente")
```

```python
# Ejercicio 6 - Leer archivo CSV
```

```python
# Leer CSV
archivo = pd.read_csv("estudiantes.csv")

print(archivo)
```
