```python
import networkx as nx
import matplotlib.pyplot as plt
import time
from collections import deque
from IPython.display import clear_output

# Crear el grafo
G = nx.Graph()

edges = [
    ('A','B'),
    ('A','C'),
    ('B','D'),
    ('B','E'),
    ('C','F'),
    ('E','F')
]

G.add_edges_from(edges)

# Posición fija para que no se mueva
pos = nx.spring_layout(G, seed=42)
```

```python
def bfs_animado(G, start):

    visited = []
    queue = deque([start])

    while queue:
        node = queue.popleft()

        if node not in visited:
            visited.append(node)

            clear_output(wait=True)

            colors = []
            for n in G.nodes():
                if n == node:
                    colors.append("orange")   # nodo actual
                elif n in visited:
                    colors.append("green")    # visitado
                else:
                    colors.append("lightblue")

            nx.draw(G, pos, with_labels=True, node_color=colors, node_size=2000)
            plt.title(f"BFS visitando: {node}")
            plt.show()

            time.sleep(1)

            queue.extend(set(G.neighbors(node)) - set(visited))
```

```python
bfs_animado(G,'A')
```

```python
def dfs_animado(G, node, visited=None):

    if visited is None:
        visited = []

    visited.append(node)

    clear_output(wait=True)

    colors = []
    for n in G.nodes():
        if n == node:
            colors.append("red")      # nodo actual
        elif n in visited:
            colors.append("green")
        else:
            colors.append("lightblue")

    nx.draw(G, pos, with_labels=True, node_color=colors, node_size=2000)
    plt.title(f"DFS visitando: {node}")
    plt.show()

    time.sleep(1)

    for neighbor in G.neighbors(node):
        if neighbor not in visited:
            dfs_animado(G, neighbor, visited)

    return visited
```

```python
dfs_animado(G,'A')
```
